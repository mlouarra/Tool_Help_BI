```python
path_file = '/home/ml/BI_Help/data/raw/Import/corresp_Labels/MATKL T023T ALL AQ6_m_1.csv'
path_file_fr ='/home/ml/BI_Help/data/processed/codes_labels_fr.pkl'
path_code_corrige = '/home/ml/BI_Help/data/processed/codes_corrige.pkl'
path_file_fr_cleaned = '/home/ml/BI_Help/data/processed/codes_labels_fr_cleaned.pkl'
path_code_to_remove = '/home/ml/BI_Help/data/processed/code_to_remove.pkl'
```

```python
import pandas as pd
import re
```

```python
pattern_1 = re.compile('ne plus|pas utiliser', re.IGNORECASE)
pattern_2 = re.compile('utiliser', re.IGNORECASE) 
```

```python
df = pd.read_csv(path_file, sep='|', encoding='ISO-8859-15')
```

```python
df['Language'] = df['Language'].str.strip()
```

```python
df_fr = df[df['Language'] == 'FR']
```

```python
df_fr.dropna(inplace=True)
```

```python
df_fr['Matl Group'] = df_fr['Matl Group'].str.strip()
```

```python
df_fr.to_pickle(path_file_fr)
```

```python
df_read = pd.read_pickle(path_file_fr)
```

```python
df_remove_code = df_read[df_read['Material grp desc. 2'].str.contains(pat = pattern_1, regex=True )]
```

```python
df_remove_code.to_pickle(path_code_to_remove)
```

```python
df_remove_code
```

```python
df_read = df_read[~df_read['Material grp desc. 2'].str.contains(pat=pattern_1, regex=True )]
```

```python
df_read.to_pickle(path_file_fr_cleaned)
```

```python
df_list_to_remove = list(df_remove_code['Matl Group'])
```

```python
df_read = df_read[~df_read['Matl Group'].isin(df_list_to_remove)]
```

```python
df_read.sample(20)
```

```python
def df_code_correction():
    df = pd.read_pickle(path_file_fr_cleaned)
    df = df[df['Material grp desc. 2'].str.contains(pat = pattern_2, regex=True)]
    df['code_correction'] = df['Material grp desc. 2'].apply(lambda x: pattern_2.sub('', x))
    df['code_correction'] = df['code_correction'].str.strip()
    return df
```

```python
df_code_corrige = df_code_correction()
```

```python
df_code_corrige.drop_duplicates(inplace=True)
```

```python
df_code_corrige[['Matl Group', 'code_correction']].to_pickle(path_code_corrige)
```

```python
df = pd.read_pickle(path_code_corrige)
```
